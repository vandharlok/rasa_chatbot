```python
def decode_unicode_escapes(text):
    try:
        text = text.encode('latin1', errors='ignore').decode('utf-8')
        return text
    except (AttributeError, UnicodeDecodeError):
        return text

def extract_data(row):
    try:
        data = json.loads(row['data'])
    except json.JSONDecodeError as e:
        print(f"Erro ao decodificar JSON: {e}")
        return None

    if row['type_name'] in ['user', 'bot']:
        return {
            'type_name': row['type_name'],
            'text': decode_unicode_escapes(data.get('text', '')),
            'timestamp': data.get('timestamp')
        }
    elif row['type_name'] == 'slot':
        slot_name = data.get('name')
        slot_value = data.get('value')
        if slot_name == 'requested_slot' or slot_value is None:
            return None
        return {
            'type_name': row['type_name'],
            'slot_name': slot_name,
            'slot_value': slot_value,
            'timestamp': data.get('timestamp')
        }
    return None

def optimized_version(df_chatbot):
    # Verificando se o DataFrame está vazio
    if df_chatbot.empty:
        print("O DataFrame está vazio.")
        return
    
    # Faça uma cópia explícita para evitar o SettingWithCopyWarning
    df_chatbot = df_chatbot.copy()

    # Desloca as colunas para comparar com a próxima linha
    df_chatbot.loc[:, 'next_type_name'] = df_chatbot['type_name'].shift(-1)
    df_chatbot.loc[:, 'next_action_name'] = df_chatbot['action_name'].shift(-1)

    # Filtra as intenções onde o tipo da linha atual é 'user' e a próxima linha é a ação correspondente
    filtered_intents = df_chatbot[
        (df_chatbot['type_name'] == 'user') &
        (df_chatbot['next_type_name'] == 'action') &
        (df_chatbot['next_action_name'] == df_chatbot['intent_name'].apply(lambda x: f'utter_{x}'))
    ]['intent_name'].tolist()

    # Processa e extrai os dados
    df_chatbot.loc[:, 'extracted_data'] = df_chatbot.apply(extract_data, axis=1)
    
    # Inicializa o defaultdict para armazenar dados de sessão
    session_data = defaultdict(lambda: {'messages': [], 'slots': defaultdict(list), 'intent_names': set()})

    # Itera sobre o dataframe para processar as linhas
    for _, row in df_chatbot.iterrows():
        session_id = row['sender_id']
        extracted_data = row['extracted_data']
        
        if extracted_data is None:
            continue

        if row['type_name'] in ['user', 'bot']:
            session_data[session_id]['messages'].append(extracted_data)
        elif row['type_name'] == 'slot':
            slot_name = extracted_data['slot_name']
            slot_value = extracted_data['slot_value']
            if slot_value not in session_data[session_id]['slots'][slot_name]:
                session_data[session_id]['slots'][slot_name].append(slot_value)
            session_data[session_id]['messages'].append({
                'type_name': 'slot',
                'slot_name': slot_name,
                'slot_value': slot_value,
                'timestamp': extracted_data['timestamp']
            })
        
        if row['intent_name'] in filtered_intents:
            session_data[session_id]['intent_names'].add(row['intent_name'])

    # Verifica se `session_data` foi populado
    if not session_data:
        print("Nenhum dado de sessão foi processado.")
        return

    # Ordenar mensagens e processar os dados finais
    for session_id in session_data:
        session_data[session_id]['messages'].sort(key=lambda x: x['timestamp'])
        session_data[session_id]['intent_names'] = list(session_data[session_id]['intent_names'])
        session_data[session_id]['slots'] = {key: value for key, value in session_data[session_id]['slots'].items() if value}

    # Converter dados de sessão para JSON e salvar em arquivo
    try:
        session_data_json = json.dumps(session_data, indent=2, ensure_ascii=False)
        with open('session_data.json', 'w', encoding='utf-8') as f:
            f.write(session_data_json)
        print("session_data.json criado com sucesso.")
    except Exception as e:
        print(f"Erro ao criar session_data.json: {e}")

    # Estatísticas dos slots
    slot_statistics = defaultdict(int)

    for session in session_data.values():
        for slot_name, slot_list in session['slots'].items():
            slot_statistics[slot_name] += len(slot_list)

    # Converter estatísticas de slots para JSON e salvar em arquivo
    try:
        slot_statistics_json = json.dumps(slot_statistics, indent=2, ensure_ascii=False)
        with open('slot_statistics.json', 'w', encoding='utf-8') as f:
            f.write(slot_statistics_json)
        print("slot_statistics.json criado com sucesso.")
    except Exception as e:
        print(f"Erro ao criar slot_statistics.json: {e}")

# Testando a versão otimizada
optimized_version(df_chatbot.copy())
```

```python
import time
import json
from collections import defaultdict
import pandas as pd
from sqlalchemy import create_engine


# Função para obter o DataFrame atualizado do banco de dados
def get_updated_dataframe():
    engine = create_engine(f'postgresql+psycopg2://{PG_USER}:{PG_PASSWORD}@localhost:{PG_PORT}/{PG_DATABASE}')
    df_chatbot = pd.read_sql_table('events', engine)
    return df_chatbot

def optimized_version_for_slots(df_chatbot):
    # Verificando se o DataFrame está vazio
    if df_chatbot.empty:
        print("O DataFrame está vazio.")
        return
    
    # Faça uma cópia explícita para evitar o SettingWithCopyWarning
    df_chatbot = df_chatbot.copy()

    # Processa e extrai os dados
    df_chatbot['extracted_data'] = df_chatbot.apply(extract_data, axis=1)
    
    # Inicializa o defaultdict para armazenar dados de sessão focando nos slots
    session_data = defaultdict(lambda: {'slots': defaultdict(list)})

    # Itera sobre o dataframe para processar as linhas
    for _, row in df_chatbot.iterrows():
        session_id = row['sender_id']
        extracted_data = row['extracted_data']
        
        if extracted_data is None:
            continue

        if row['type_name'] == 'slot':
            slot_name = extracted_data['slot_name']
            slot_value = extracted_data['slot_value']
            if slot_value not in session_data[session_id]['slots'][slot_name]:
                session_data[session_id]['slots'][slot_name].append(slot_value)

    # Verifica se `session_data` foi populado
    if not session_data:
        print("Nenhum dado de sessão foi processado.")
        return

    # Ordenar e processar os dados finais
    for session_id in session_data:
        session_data[session_id]['slots'] = {key: value for key, value in session_data[session_id]['slots'].items() if value}

    # Verificar se há mudanças nos dados e atualizar o JSON
    try:
        with open('session_data.json', 'r', encoding='utf-8') as f:
            previous_data = json.load(f)
    except FileNotFoundError:
        previous_data = {}

    if session_data != previous_data:
        session_data_json = json.dumps(session_data, indent=2, ensure_ascii=False)
        with open('session_data.json', 'w', encoding='utf-8') as f:
            f.write(session_data_json)
        print("session_data.json atualizado com sucesso.")
    else:
        print("Nenhuma atualização nos slots detectada.")

    # Estatísticas dos slots
    slot_statistics = defaultdict(int)

    for session in session_data.values():
        for slot_name, slot_list in session['slots'].items():
            slot_statistics[slot_name] += len(slot_list)

    # Verificar se há mudanças nas estatísticas e atualizar o JSON
    try:
        with open('slot_statistics.json', 'r', encoding='utf-8') as f:
            previous_statistics = json.load(f)
    except FileNotFoundError:
        previous_statistics = {}

    if slot_statistics != previous_statistics:
        slot_statistics_json = json.dumps(slot_statistics, indent=2, ensure_ascii=False)
        with open('slot_statistics.json', 'w', encoding='utf-8') as f:
            f.write(slot_statistics_json)
        print("slot_statistics.json atualizado com sucesso.")
    else:
        print("Nenhuma atualização nas estatísticas dos slots detectada.")

# Loop infinito para verificar atualizações a cada 1 segundo
while True:
    df_chatbot = get_updated_dataframe()  # Obtém os dados mais recentes do banco de dados
    optimized_version_for_slots(df_chatbot)
    time.sleep(1)  # Espera 1 segundo antes de rodar novamente
```
